```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
```

```python
df = pd.read_parquet("hf://datasets/julien040/hacker-news-posts/story.parquet")
```

```python
df.head()
```

```python
titles = df["title"].tolist() # only taking titles
```

```python
len_titles = np.array([len(title.split()) for title in titles]) # count words in titles
```

```python
len_titles
```

Compute and plot histogram

```python
hist_counts, bin_edges = np.histogram(len_titles, bins=30)
plt.hist(len_titles, bins=30, edgecolor='black')
plt.xlabel('Word Count')
plt.ylabel('Frequency')
plt.show()
```
