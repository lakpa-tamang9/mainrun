```python
import os
import json
import matplotlib.pyplot as plt
```

Plotting the learning curves of the trained model

```python
def get_logs_and_plot(root_dir):
    files = os.listdir(root_dir)
    runs = {}
    for file in files:
        if not file.endswith('.log'):
            continue
        filename = file.split('.')[0]
        val_losses = []
        with open(os.path.join(root_dir, file), "r") as f:
            logs = [json.loads(line) for line in f]
        val_losses.extend(log["loss"] for log in logs if log["event"] == "validation_step")
        runs[filename] = val_losses
    plt.figure(figsize=(6, 4))
    for run, losses in runs.items():
        plt.plot(losses, linestyle='--', label=f'{run}')
    plt.xticks(range(0, len(val_losses), 3))
    plt.legend(loc="upper right", fontsize=9, title_fontsize=10)
    plt.ylim(0,0.2)
    plt.xlabel("Validation steps", fontsize=12)
    plt.ylabel("Validation Loss", fontsize=12)
    plt.tight_layout()
    plt.show()
    
```

```python
get_logs_and_plot("/home/tamangld/llm/mainrun/mainrun/logs2")
```
